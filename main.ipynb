### Process data to get analysed data (distance, velocity, acceleration, jitter, distance to surface, tail-to-head angle, more WIP)

This step will take the DLC csv outputs and do:
1. data integrity checks (null + low threshold + jitter) using `data_integrity.py`
2. convert pixels to mm
3. calculate distance, velocity and acc using `analysis.py`

All of these are run using the one script, which is `data_mm_analyse.py`

The critical flags are `--csv_path` to give DLC csv path if not ./data/csv/, `--output_path` to give output path if not ./data/output/mm_analyse, `--invalid_thresh` to set the likelihood threshold to be considered invalid if not 0.25, and `--percent_missing` to set the percentage threshold for considering a csv "problematic" if not 15%.

A problems_csv.txt is also saved to output path, collating all the issues with the csvs for a detailed look.
